# Data Exploration
Check class balance and preview sample images before training.

```python
import pathlib
import matplotlib.pyplot as plt
from PIL import Image

DATA_DIR = pathlib.Path('../data/raw')
```

```python
# Count images per class
class_counts = {}
for class_dir in sorted(DATA_DIR.iterdir()):
    if class_dir.is_dir():
        n = len(list(class_dir.glob('*')))
        class_counts[class_dir.name] = n

for name, count in class_counts.items():
    print(f'{name}: {count}')
```

```python
plt.figure(figsize=(10, 4))
plt.bar(class_counts.keys(), class_counts.values())
plt.xticks(rotation=45, ha='right')
plt.ylabel('image count')
plt.title('Class balance')
plt.tight_layout()
plt.show()
```

```python
# Preview a few samples per class
classes = sorted(class_counts.keys())
fig, axes = plt.subplots(len(classes), 4, figsize=(12, 3 * len(classes)))
for row, class_name in enumerate(classes):
    samples = list((DATA_DIR / class_name).glob('*'))[:4]
    for col, img_path in enumerate(samples):
        ax = axes[row, col] if len(classes) > 1 else axes[col]
        img = Image.open(img_path)
        ax.imshow(img)
        ax.set_title(class_name, fontsize=8)
        ax.axis('off')
plt.tight_layout()
plt.show()
```
